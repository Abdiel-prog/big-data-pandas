```python
import pandas as pd
import numpy as np
import glob
```

```python
orders_files = sorted(glob.glob("data/orders_2024_*.csv"))
print(orders_files)

df_orders_raw = pd.concat(
    [pd.read_csv(f) for f in orders_files],
    ignore_index=True
)

print(df_orders_raw.info(memory_usage="deep"))
```

```python
orders_dtypes = {
    "order_id": "int32",
    "user_id": "int32",
    "product_id": "int16",
    "price": "float32",
    "quantity": "int16",
}

df_orders_opt = pd.concat(
    [
        pd.read_csv(f, dtype=orders_dtypes, parse_dates=["order_datetime"])
        for f in orders_files
    ],
    ignore_index=True
)

df_orders_opt["status"] = df_orders_opt["status"].astype("category")

print(df_orders_opt.info(memory_usage="deep"))
```
